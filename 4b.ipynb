```python

import pandas as pd
import re

from gensim.models import Word2Vec
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# 1. Load dataset
data = pd.read_csv('twitter.csv')

# 2. Preprocess text
stop_words = set(ENGLISH_STOP_WORDS)

def clean_text(text):
    text = re.sub('[^a-zA-Z]', ' ', str(text))
    words = text.lower().split()
    words = [w for w in words if w not in stop_words]
    return words

sentences = data['text'].dropna().apply(clean_text).tolist()

# 3. Train CBOW model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,
    workers=4
)

# 4. Evaluate results
print("Vocabulary:", list(model.wv.index_to_key[:10]))

word = 'good'

if word in model.wv:
    print("Similar words:", model.wv.most_similar(word))
else:
    print("Word not found in vocabulary")

# 5. Get word embedding
print("Word Vector:", model.wv[word] if word in model.wv else "Not found")
```
