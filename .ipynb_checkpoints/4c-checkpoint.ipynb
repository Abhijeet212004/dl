```python

import re

from gensim.models import Word2Vec
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# 1. Load document
with open('document3.txt', 'r', encoding='utf-8') as file:
    text = file.read()

# 2. Preprocess text
stop_words = set(ENGLISH_STOP_WORDS)

def clean_text(text):
    text = re.sub('[^a-zA-Z]', ' ', text)
    words = text.lower().split()
    words = [w for w in words if w not in stop_words]
    return words

sentences = [clean_text(text)]

# 3. Train CBOW model
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=5,
    min_count=1,
    sg=0,
    workers=4
)

# 4. Evaluate results
print("Vocabulary:", list(model.wv.index_to_key))

word = model.wv.index_to_key[0]

print("Selected word:", word)
print("Word Vector:", model.wv[word])

print("Similar words:", model.wv.most_similar(word))
```
